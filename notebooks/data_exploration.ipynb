```python
import json
import os

path = "../data/raw/All_Beauty.jsonl"

print("Exists:", os.path.exists(path))
print("Size:", round(os.path.getsize(path) / (1024**2), 2), "MB")

with open(path, "r", encoding="utf-8") as f:
    record = json.loads(f.readline())

print("\nFirst record:")
print(record)

print("\nFields:")
for key in record:
    print("-", key)
```

```python
meta_path = "../data/raw/meta_All_Beauty.jsonl"

print("Exists:", os.path.exists(meta_path))
print("Size:", round(os.path.getsize(meta_path) / (1024**2), 2), "MB")

with open(meta_path, "r", encoding="utf-8") as f:
    meta_record = json.loads(f.readline())

print("\nFirst metadata record:")
print(meta_record)

print("\nFields:")
for key in meta_record:
    print("-", key)
```
